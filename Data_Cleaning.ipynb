#### `2:-`Now we will Handle the Missing Values 

```python
# Before we handle the missing values lets see if the missing values coming from the  customers who do not have a loan.
pd.crosstab(
    df["has_loan"],
    df["loan_type"].isnull(),
    margins=True
)
```



#### Important Observations.
- we found that 357,712 records have no loan type mentioned (which generally means that they dosenot have a lone).
- But we found that  19,422 records out of 357,712 records have a lone which is mention in `has_loan` column but in  the `loan_type` column we found that it has no lone (means these are missing value).
- Now we can say that there are 19,422 missing values in the `lone_type`column


```python
# Now we are just verifying if the column has_loan has genuine values or not .
# and to do that we are checking the emi_amount column so that we can verify if they have an active emi which can confrom that they genually have a loan or not.
df[
    (df["has_loan"] == 1) &
    (df["loan_type"].isnull()) &
    (df["emi_amount"] > 0)
][["customer_id", "has_loan", "loan_type", "emi_amount"]].sample(10)
```

```python
# Now we are creating an new category for the missing values which has a loan but the loan type is missinng
df_clean = df.copy()

df_clean.loc[
    (df_clean["has_loan"] == 1) &
    (df_clean["loan_type"].isnull()),
    "loan_type"
] = "Loan Type Unknown"

print(df_clean["loan_type"].value_counts())
```

```python
# now we are creating an new category for the missing values which dosenot have an loan
df_clean.loc[
    (df_clean["has_loan"] == 0) &
    (df_clean["loan_type"].isnull()),
    "loan_type"
] = "No Loan"
print(df_clean["loan_type"].value_counts())
```

```python
# just checking is all the missing values are gone or not
df_clean.isnull().sum()
```


#### Important Observation
- We replaced missing loan types for customers without a loan with `No Loan`.
- and we replaced missing loan types for customers with a loan with `Loan Type Unknown`.
- so that we can Preserve all transaction records without deleting any rows.


#### `3:-`Now we will check the data typs and convert them in correct format

```python
df_clean.info()
```


### Important Observations
- `transaction_date` column is stored as object and we need to convert it to a date format.
- `transaction_time` is also stored as text so we need to change that to.
- Transaction amounts, account balances, and EMI amounts are already in numeric format.
- The `has_loan` and `is_fraud` columns are stored as integers.


```python
# now we will convert the transaction_date from object to datetime format
df_clean["transaction_date"] = pd.to_datetime(
    df_clean["transaction_date"],
    errors="coerce"
)
```

```python
# now we will convert the  transaction_time from objectto a proper time format
df_clean["transaction_time"] = pd.to_datetime(
    df_clean["transaction_time"],
    format="%H:%M",
    errors="coerce"
)
```
