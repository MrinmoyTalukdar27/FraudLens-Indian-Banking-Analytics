```python
# Just adding the importent library
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
#loding the dataset
df = pd.read_csv(r"P:\Mini_Project\Datasets\Transaction_risk_insights_cleaned.csv")
df["transaction_date"] = pd.to_datetime(df["transaction_date"])
df.sample(10)
```

#### `1:-` First we extract useful information from the transaction_date column.

```python
# now we will extract year
df["transaction_year"] = df["transaction_date"].dt.year

# now we will extract month
df["transaction_month"] = df["transaction_date"].dt.month

# now we will extract day
df["transaction_day"] = df["transaction_date"].dt.day

# now we will extract day name
df["day_of_week"] = df["transaction_date"].dt.day_name()

# now we will identify weekend transactions
df["is_weekend"] = (
    df["transaction_date"].dt.dayofweek >= 5
).astype(int)
```

```python
# now we will see newly created features
df[[
    "transaction_date",
    "transaction_year",
    "transaction_month",
    "transaction_day",
    "day_of_week",
    "is_weekend"
]].sample(10)
```

#### `2:-` Now we will  do transaction time period categorization
- We already have a transaction_hour column. We'll use it to divide transactions into four time periods.

```python
# Defining the time periods
conditions = [
    (df["transaction_hour"] >= 0) & (df["transaction_hour"] < 6),
    (df["transaction_hour"] >= 6) & (df["transaction_hour"] < 12),
    (df["transaction_hour"] >= 12) & (df["transaction_hour"] < 18),
    (df["transaction_hour"] >= 18) & (df["transaction_hour"] <= 23)
]

periods = ["Night", "Morning", "Afternoon", "Evening"]

# Creating the the new feature
df["time_period"] = np.select(conditions, periods, default="Unknown")
```

```python
# Now we are checking the newly created feature
df[["transaction_hour", "time_period"]].sample(10)
```

#### `3:-` Now we will do transaction Amount Categorization

```python
# Now we will make three catagories based on transaction amount like low, medium and high
conditions = [
    df["transaction_amount"] < 10000,
    (df["transaction_amount"] >= 10000) &
    (df["transaction_amount"] <= 50000),
    df["transaction_amount"] > 50000
]

categories = ["Low", "Medium", "High"]

df["transaction_amount_category"] = np.select(
    conditions,
    categories,
    default="Unknown"
)
```

```python
# Now we will see the newly created feature
df[["transaction_amount","transaction_amount_category"]].head(10)
```

```python
df["transaction_amount_category"].value_counts()
```

```python
# now we will calculate the percentage of each category in transaction_amount_category
(df["transaction_amount_category"].value_counts(normalize=True) * 100).round(2)
```

#### `4:-` Now we will do the Account Balance Categorization

```python
# We will categorize account balances into three groups.
conditions = [
    df["account_balance"] <= 25000,
    (df["account_balance"] > 25000) &
    (df["account_balance"] <= 100000),
    df["account_balance"] > 100000
]

categories = ["Low", "Medium", "High"]

df["balance_category"] = np.select(
    conditions,
    categories,
    default="Unknown"
)
```

```python
df[["account_balance", "balance_category"]].head(10)
```

```python
# now we will verify the balance category distribution
df["balance_category"].value_counts()
# now we will calculate the percentage of each category in balance_category
(df["balance_category"].value_counts(normalize=True) * 100).round(2)
```

#### `5:-` Now we will do credit Score Categorization

```python
# Now we will categorize credit scores into three groups.
conditions = [
    (df["credit_score"] >= 300) & (df["credit_score"] <= 499),
    (df["credit_score"] >= 500) & (df["credit_score"] <= 699),
    (df["credit_score"] >= 700) & (df["credit_score"] <= 899)
]

categories = ["Low", "Medium", "High"]

df["credit_score_category"] = np.select(
    conditions,
    categories,
    default="Unknown"
)
```

```python
df[["credit_score", "credit_score_category"]].sample(10)
```

```python
# Now we will verify the category distribution
df["credit_score_category"].value_counts()
```

```python
# now we will calculate Category percentages
(df["credit_score_category"].value_counts(normalize=True) * 100).round(2)
```

#### `6:-` Now we will make transaction frequency Feature

```python
# This feature will calculate how many transactions each customer has made in the dataset.

# Count the total transactions made by each customer
df["customer_transaction_count"] =(
    df.groupby("customer_id")["transaction_id"]
    .transform("count")
)
```

```python
df[["customer_id","transaction_id","customer_transaction_count"]].sample(10)
```
